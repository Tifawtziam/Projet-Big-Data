```python
from pyspark.sql import functions as F

storage_account = "energybigdatastorage"

# LECTURE depuis processed
df = spark.read.format("delta").load(
    f"abfss://processed@{storage_account}.dfs.core.windows.net/informations_households/")

# Trim + Upper sur LCLid
df = df.withColumn("LCLid", F.upper(F.trim(F.col("LCLid"))))

# SAUVEGARDE en écrasant
df.write.format("delta").mode("overwrite") \
    .save(f"abfss://processed@{storage_account}.dfs.core.windows.net/informations_households/")
print(" informations_households mis à jour !")
```
