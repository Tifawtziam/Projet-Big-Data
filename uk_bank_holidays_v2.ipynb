```python
from pyspark.sql import functions as F

storage_account = "energybigdatastorage"

# LECTURE depuis processed
df = spark.read.format("delta").load(
    f"abfss://processed@{storage_account}.dfs.core.windows.net/uk_bank_holidays/")

# Ajouter colonne year
df = df.withColumn("year", F.year(F.col("Bank_holidays")))

# Vérification
df.show()

# SAUVEGARDE en écrasant avec nouvelle schema
df.write.format("delta").mode("overwrite") \
    .option("overwriteSchema", "true") \
    .save(f"abfss://processed@{storage_account}.dfs.core.windows.net/uk_bank_holidays/")
print(" uk_bank_holidays mis à jour !")
```
